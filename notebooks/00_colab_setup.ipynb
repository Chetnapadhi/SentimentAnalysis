## What it does
1. Clones the GitHub repository (or pulls latest changes)
2. Installs `requirements.txt`
3. Verifies CUDA / GPU availability
4. Sets the repository as the working directory
5. Prints commands to run E0 / E1

**Does NOT auto-train.** Run experiments explicitly from the provided commands.

````python
# Cell 1: Clone / pull the repository
REPO_URL = "https://github.com/Chetnapadhi/SentimentAnalysis"
REPO_DIR = "SentimentAnalysis"

import os, subprocess, sys

if not os.path.exists(REPO_DIR):
    print(f"Cloning {REPO_URL}...")
    subprocess.run(["git", "clone", REPO_URL], check=True)
else:
    print(f"Repository exists, pulling latest...")
    os.chdir(REPO_DIR)
    subprocess.run(["git", "pull"], check=True)

os.chdir(REPO_DIR)
print(f"Working directory: {os.getcwd()}")
````

````python
# Cell 3: Verify CUDA / GPU
import torch
print("CUDA available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("CUDA version:", torch.version.cuda)
    print("Memory:", round(torch.cuda.get_device_properties(0).total_memory / 1e9, 1), "GB")
else:
    print("WARNING: No GPU detected. Switch runtime to T4 GPU: Runtime -> Change runtime type -> T4 GPU")
````
